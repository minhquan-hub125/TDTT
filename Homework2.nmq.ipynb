Bài 8: Giải hệ phương trình tuyến tính:
ax + by = m
cx + dy = n

```python
a,b,m = map(float, input().split())
c,d,n = map(float, input().split())
D = a*d - b*c
Dx = m*d - b*n
Dy= a*n - m*c
if D!=0:
    x = Dx/D
    y = Dy/D
    print(f"He phuong trinh co nghiem duy nhat: x = {x}, y = {y}")
elif D == 0 and Dx == 0 and Dy == 0:
    print("He phuong trinh co vo so nghiem")
elif D == 0 and Dx != 0 or Dy != 0:
    print("He phuong trinh ko co nghiem")
```

Bài 10: Nhập vào tâm và bán kính của một đường tròn. Sau đó nhập vàomộtđiểm A(x, y) bất kì và kiểm tra xem nó có thuộc đường tròn hay không?
